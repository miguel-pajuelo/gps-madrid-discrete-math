```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G = nx.Graph()
G.add_node(1)
G.add_node(2)
G.add_edge(1,2)
G.add_nodes_from([3,4,5])
G.add_edges_from([(1,4),(3,4),(2,4),(3,2),(3,5)])
print(G)
print(G.nodes)
print(G.edges)
```

```python
G.remove_node(5)
print(G)
print(G.nodes)
print(G.edges)
```

```python
plot = plt.plot()
nx.draw(G,with_labels=True,pos={1:(0,0),2:(0,1),3:(1,0),4:(1,1)})
plt.show()
```

```python
G2 = nx.Graph()
G2.add_nodes_from(["a","b",10,11,3.14])
G2.add_edge("a",10,object="Una cadena",weight=1.1)
G2.add_edges_from([("a",3.14,{"object":"Otra cadena","weight":1.2})])
```

# Digraph

```python
Gd=nx.DiGraph()
Gd.add_nodes_from([1,2,3,4])
Gd.add_edges_from([(1,2),(2,1),(1,4),(3,4),(2,4),(4,2),(3,2)])
plot = plt.plot()
nx.draw(Gd,with_labels=True,)
plt.show()
```
